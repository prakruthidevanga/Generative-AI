## Objective
The objective of this assignment is to understand how different text-generation parameters influence the output produced by a pretrained language model. Specifically, this notebook explores how **temperature**, **top-k**, **top-p**, and **repetition penalty** change the creativity, randomness, predictability, and repetitiveness of text generated by GPT-2. By keeping the same opening sentence and only changing one parameter at a time, it becomes possible to isolate the effect each parameter has on the output, rather than seeing several effects mixed together.


## Introduction

GPT-2 is a transformer-based language model released by OpenAI that generates text by predicting the next word (token) in a sequence, one token at a time, based on everything that came before it. At each step, the model produces a probability distribution over its entire vocabulary, and a *decoding strategy* decides which token actually gets picked.

This assignment tests four such strategies:

- **Temperature** — scales confidence: lower = sharper/safer, higher = flatter/more random.
- **Top-k** — only samples from the *k* most likely next tokens.
- **Top-p** — samples from the smallest set of tokens covering *p* cumulative probability.
- **Repetition penalty** — lowers the odds of reusing already-generated tokens.

the pretrained GPT-2 model from the Hugging Face **transformers library** is used to generate text from one opening sentence, five times, changing only one generation parameter each time. A custom prompt is then generated three different ways, and the results are compared and analysed.

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch
```

## 2. Loading the GPT-2 Model

```python
model_name = "gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

model.eval()

print("GPT-2 model loaded successfully!")
```

## 3. Creating a Text Generation Function

```python
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0,
                   repetition_penalty=1.0, max_length=100):
    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=max_length,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=repetition_penalty,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return generated_text
```

## 4. Part 1 — Five Experiments on One Opening Sentence

**Opening sentence used for all 5 experiments below** (this is the same prompt across every experiment — only one parameter changes each time, as required by the assignment).

```python
prompt = "The projects focuses on solving a real-world problem using technology (AI)."
```

### Experiment 1 — Parameter Changed: Temperature (value = 0.5)

```python
output1 = generate_text(
    prompt,
    temperature=0.5,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print(output1)
```

**Observation:** A low *temperature (0.5)* makes the model favour the most probable next word almost every time. The output reads as fairly safe, grammatically correct, and somewhat predictable/repetitive.

### Experiment 2 — Parameter Changed: Top-k (value = 10)

```python
output2 = generate_text(
    prompt,
    temperature=1.0,
    top_k=10,
    top_p=1.0,
    repetition_penalty=1.0
)

print(output2)
```

**Observation:** With *top_k=10*, the model can only sample from the 10 most likely next words at each step. This narrows the vocabulary choice, producing coherent but slightly repetitive text — less diverse than unrestricted sampling.

### Experiment 3 — Parameter Changed: Top-p / Nucleus Sampling (value = 0.7)

```python
output3 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=0.7,
    repetition_penalty=1.0
)

print(output3)
```

**Observation:** *top_p=0.7* restricts sampling to the smallest set of words whose cumulative probability is 70%. This gives a good balance — more natural variety than low top-k, while staying mostly coherent.

### Experiment 4 — Parameter Changed: Repetition Penalty (value = 1.5)

```python
output4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.5
)

print(output4)
```

**Observation:** GPT-2 has a known tendency to fall into loops, repeating the same word or short phrase over and over once it gets 'stuck'. Setting

*repetition_penalty=1.5* visibly reduces this — tokens that already appeared earlier in the generated text get their probability lowered, so the model is pushed toward introducing new vocabulary instead of reusing what it just said.


### Experiment 5 — Parameter Changed: Temperature (value = 1.5, high)

```python
output5 = generate_text(
    prompt,
    temperature=1.5,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print(output5)
```

**Observation:** This is the opposite of Experiment 1. *higher temperature (1.5)*  flattens the probability distribution, so tokens that would normally be very unlikely get picked far more often. The result reads as noticeably more 'creative' and unpredictable than any of the other four experiments, but it also starts losing grammatical structure toward the end of the sentence word choices feel more associative than logical.



## 5. Additional Activity — Extreme Temperature (Unusual / Meaningless Output)

Pushing temperature very high (e.g. 3.0) flattens the probability distribution almost completely, so the model samples close to uniformly at random over the vocabulary. This experiment intentionally tries to trigger the kind of unusual/meaningless output mentioned in the Additional Activity.

```python
extreme_output = generate_text(
    prompt,
    temperature=3.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print(extreme_output)
```

**Observation:** At *temperature=3.0* the output stops looking like English sentences at all — it turns into a string of loosely connected or completely unrelated words, sometimes mixing in odd punctuation, fragments of unrelated topics, or repeated symbols. It's clearly 'broken' compared to every other experiment.

**Explanation of why this happens:** Temperature scales the logits before the softmax step. At *temperature=3.0*, the differences between high- and low-probability tokens are compressed, so unlikely, unrelated tokens become almost as likely to be picked as sensible ones. The model ends up chaining together words that don't logically follow from each other, producing incoherent or nonsensical text.


## 6. Part 2 — Custom Prompt: Three Generations

A short custom prompt of my own (a **product description**) is generated three different ways below, as required by the assignment.

```python
custom_prompt = "It started with a simple idea."
```

**Observation:** The repetition penalty keeps the description from looping on the same phrase.

## 8. Analysis

**Which settings produced more repetitive / predictable text?**

Experiment 1 (low temperature = 0.5) and Experiment 2 (low top-k = 10) — both narrow the model's choices to the most probable next words, so the output stays safe and can loop or repeat.

**Which settings produced more creative text?**

Experiment 3 (top-p = 0.7) and Experiment 4 (repetition penalty = 1.5) — these add controlled diversity without letting the output fall apart grammatically.

**Which settings produced more random / incoherent text?**

Experiment 5 (temperature = 1.5) and especially the Additional Activity (temperature = 3.0) — high temperature flattens the sampling distribution, so unlikely and unrelated words get picked, breaking coherence.

**Overall pattern:** temperature, top-k, and top-p all control *how much of the probability distribution the model is allowed to sample from*, just in different ways (scaling, fixed count, and cumulative probability respectively), while repetition penalty is a separate mechanism that specifically targets already-used tokens rather than reshaping the distribution as a whole. Understanding this distinction helped explain why, e.g., Experiment 4 could reduce repetition without making the text noticeably more 'random' the way raising temperature did.

## 9. Conclusion

This assignment showed that GPT-2's output quality and style depend heavily on the decoding strategy used at generation time, not just on the model itself. Lower temperature and lower top-k values push the model toward safe, repetitive text; top-p sampling and repetition penalty offer more controlled ways to introduce variety; and pushing temperature too high breaks the output down into incoherent noise. In practice, this suggests that for tasks like factual answers or code generation, lower temperature / top-k settings are preferable, while for creative writing tasks, a moderate top-p value combined with a repetition penalty gives the best mix of originality and readability.
